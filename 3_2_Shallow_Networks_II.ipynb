```julia
# Draw a 2D function: filled contour with line contour ovaerlaid 
function draw_2D_function(x₁_mesh, x₂_mesh, y; title="")
    p = contourf(x₁_mesh, x₂_mesh, y; levels=256, c=:hot, clims=(-10, 10),
        xlabel="x₁", ylabel="x₂", title=title)
    contour!(p, x₁_mesh, x₂_mesh, y; levels=-10:1.0:10, c=:winter, colorbar=false)
    return p 
end 
```
