```julia
using Plots
```

```julia
# Draw a 2D function: filled contour with line contour ovaerlaid 
function draw_2D_function(x₁_mesh, x₂_mesh, y; title="")
    p = contourf(x₁_mesh, x₂_mesh, y; levels=256, c=:hot, clims=(-10, 10),
        xlabel="x₁", ylabel="x₂", title=title)
    contour!(p, x₁_mesh, x₂_mesh, y; levels=-10:1.0:10, c=:winter, colorbar=false)
    return p 
end 

# Plot the shallow neural network. Input range [0,10]x[0,10], output [-10, 10]
function plot_neural_2_inputs(x₁, x₂, y, pre_1, pre_2, pre_3, act_1, act_2, act_3, w_act_1, w_act_2, w_act_3)
    p₁ = draw_2D_function(x₁, x₂, pre_1; title="Preactivation")
    p₂ = draw_2D_function(x₁, x₂, pre_2; title="Preactivation")
    p₃ = draw_2D_function(x₁, x₂, pre_3; title="Preactivation")
    p₄ = draw_2D_function(x₁, x₂, act_1; title="Activation")
    p₅ = draw_2D_function(x₁, x₂, act_2; title="Activation")
    p₆ = draw_2D_function(x₁, x₂, act_3; title="Activation")
    p₇ = draw_2D_function(x₁, x₂, w_act_1; title="Weighted Act")
    p₈ = draw_2D_function(x₁, x₂, w_act_2; title="Weighted Act")
    p₉ = draw_2D_function(x₁, x₂, w_act_3; title="Weighted Act")
    grid = plot(p₁, p₂, p₃, p₄, p₅, p₆, p₇, p₈, p₉; layout=(3,3), size=(850, 850))
    display(grid)

    out = draw_2D_function(x₁, x₂, y; title="Network Output,")
    plot!(out; aspect_ratio=1.0)
    display(out)
end
```
