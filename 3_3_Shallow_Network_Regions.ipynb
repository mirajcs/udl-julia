# Notebook 3.3 -- Shallow network regions

The number of regions $N$ created by a shallow neural network with $D_i$ inputs and $D$ hidden units is given by Zaslavsky's formula: 
$$N = \sum_{j=0}^{D_i} \binom{D}{j} = \sum_{j=0}^{D_j} \dfrac{D!}{(D-j)!j!}$$

```julia
function number_regions(Dᵢ, D)
    N = big(0) 
    for j in 0:Dᵢ
        N += binomial(big(D), j)
    end
    return N 
end
```

```julia
# Calculate the number of regions for 2D input (Dᵢ = 2) and 3 hidden units (D=3) as in the figure 3.8j
N = number_regions(2,3)
println("Dᵢ=2, D=3, Number of regions = $(Int(N)), True value = 7 ")
```

```julia
# Calculate the number of regions for 10D input (Dᵢ = 10) and 50 hidden units (D=50)
N = number_regions(10,50)
println("Dᵢ=10, D=50, Number of regions = $(Int(N)), True value = 13432735556 ")
```

This works but there is a complication. If the number of hidden units $D$ is fewer thatn the number of input dimension $D_i$, the formula will fail. When this is the case, there are just $2^D$ regions.

```julia
# Depending on how you implemented it, the calculation may fail when Dᵢ > D 
try 
    N = number_regions(10, 8)
    println("Dᵢ = 10, D=8, number of regions = $(Int(N)), True = value=256")
catch e 
    println("An exception occured: ", error)
end  
```

```julia
# Now let's plot the graph from figure 3.9a 
using Plots 

dims = [1, 5, 10, 50, 100]
regions = zeros(length(dims), 1000)
for c_dim in 1:length(dims)
    Dᵢ = dims[c_dim]
    println("Counting regions for $Dᵢ input dimensions")
    for D in 0:999
        regions[c_dim, D+1] = number_regions(min(Dᵢ, D), D)
    end 
end

plot(0:999, regions[1, :], yscale=:log10, color=:black, label="Dᵢ=1")
plot!(0:999, regions[2, :], yscale =:log10, color=:blue, label="Dᵢ=5")
plot!(0:999, regions[3, :], yscale=:log10, color=:magenta, label="Dᵢ=10")
plot!(0:999, regions[4, :], yscale=:log10, color=:cyan, label="Dᵢ=50")
plot!(0:999, regions[5, :], yscale=:log10, color=:yellow, label="Dᵢ=100")
xlabel!("Number of hidden units, D")
ylabel!("Number of regions, N")
xlims!(0, 1000)
ylims!(1e1, 1e150)
```

```julia
# Now let's compute and plot the number of regions as a function of the number of parameters as in figure 3.9b

function number_parameters(Dᵢ, D)
    D₀ = 1 # number of outputs 
    return (Dᵢ + 1) * D + (D + 1) * D₀
end

```

```julia
# let's check the code 

N = number_parameters(10, 8)
println("Dᵢ = 10, D =8, Number of parameters = $(Int(N)), True value = 97")
```

```julia
# let's plot the graph from figure 3.9b

dims = [1, 5, 10, 50, 100]
regions = zeros(length(dims), 199)
params = zeros(length(dims), 199)

for c_dim in 1:length(dims)
    Dᵢ = dims[c_dim]
    println("Counting regions for $Dᵢ input dimensions")
    for c_hidden in 1:199
        # Iterate over different ranges or number of hidden units for different input sizes
        D = Int(floor(c_hidden * 500 / Dᵢ))
        params[c_dim, c_hidden] = number_parameters(Dᵢ, D)
        regions[c_dim, c_hidden] = number_regions(min(Dᵢ, D), D)
    end 
end 

plot(params[1,:], regions[1, :], yscale=:log10, color=:black, label="Dᵢ=1")
plot!(params[2,:], regions[2, :], yscale=:log10, color=:blue, label="Dᵢ=5")
plot!(params[3,:], regions[3, :], yscale=:log10, color=:magenta, label="Dᵢ=10")
plot!(params[4,:], regions[4, :], yscale=:log10, color=:cyan, label="Dᵢ=50")
plot!(params[5,:], regions[5, :], yscale=:log10, color=:yellow, label="Dᵢ=100")

xlabel!("Number of parameters, D")
ylabel!("Number of regions, N")
xlims!(0, 100_000)
ylims!(1e1, 1e150)
```
